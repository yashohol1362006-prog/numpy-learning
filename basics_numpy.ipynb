Array Methods

```python
arr = np.array([[3, 4, 6], [2, 9, 5], [4, 3, 1]])
print(arr)
print(arr.min())
print(arr.max())
print(arr.sum())
arr = np.array([[2, 3, 5], [2, 6, 8], [1, 6, 9]])
sum_of_row = np.sum(arr,axis=1)
print(arr)
print(sum_of_row)
sum_of_column = np.sum(arr, axis=0)
print(arr)
print(sum_of_column)
mean = np.mean(arr)
print(mean)
max = arr.argmax()     #it tells on which index is max value
print(max)
min = arr.argmin()     # it tells minimum
print(min)
```

Numpy indexin and slicin

```python
arr = np.arange(1, 31)
print(arr)
print(arr[6])
print(arr[1:4])
print(arr[:6])
print(arr[4:])
arr = np.arange(1, 51).reshape(10, 5)
print(arr)
print(arr[9, 4])
print(arr[3, 2])
print(arr[9, 2])
print(arr[9, 1])
arr = np.arange(1, 51).reshape(10, 5)
print(arr)
print(arr[4, 2])
arr = np.arange(1, 51).reshape(10, 5)
print(arr)
print(arr[7, 3])
arr = np.arange(1, 51).reshape(10, 5)
print(arr)
print(arr[5,])
arr = np.arange(1, 51).reshape(5, 10)
print(arr)
print(arr[4, 9])
arr = np.arange(1, 51).reshape(5, 10)
print(arr)
print(arr[0:3, 2:7])
arr = np.arange(1, 51).reshape(10, 5)
print(arr)
print(arr[0:4, 1:4])
arr = np.arange(1, 51).reshape(5, 10)
print(arr)
print(arr[3:5, 4:11])
```

Array Operations

1) arithmetic error:

```python
a1 = np.array([1, 2, 3, 4, 5])
a2 = np.array([6, 7, 8, 9, 10])
sum = a1 + a2
print(sum)
minus = a1 -a2
print(minus)
multiply = a1 * a2
print(multiply)
devide = a1 / a2
print(devide)
power = a1** a2
print(power)
```

Broadcasting

```python
l = [10, 20, 30, 40]
arr = np.array(l)
print(arr+10)
arr = np.arange(1, 26).reshape(5, 5)
print(arr)
print(arr+10)                   #it'll add +10 in each element of matrix
```

Deep and shalow copy

```python
arr = np.arange(1, 21)
print(arr)
slice = arr[:5]
multi_to_slice = slice*5
print(multi_to_slice)
arr = np.arange(1, 51).reshape(5, 10)
print(arr)
slice = arr[0:3, 1:8]
multi_to_slice = slice*5
print(multi_to_slice)
```

```python
a = np.array([[1, 2], [3, 4]])
```

Array Stacking(in which we joint two arrays)

```python
a = np.array([1, 2, 3, 4])
```

```python
b = np.array([5, 6, 7, 8])
```

```python
vert_stack = np.vstack((a, b))
hori_stack = np.hstack((a, b))
print(vert_stack)
print(hori_stack)
#column stack
column = np.column_stack((a, b))
print(column)
#splitting arrays
```

1) horizaontal splitting

```python
c = np.arange(16).reshape(4, 4)
print(c)
splitting = np.hsplit(c, 2)              # it'll split array in two parts as mentioned
print(splitting)
spllitting_in_4 = np.hsplit(c, 4)
print(spllitting_in_4)
```

2) vertical splitting

```python
d = np.arange(16).reshape(4,4)
print(d)
splitting_2 = np.vsplit(d, 2)
print(splitting_2)
split_4 = np.vsplit(d, 4)
print(split_4)
```
